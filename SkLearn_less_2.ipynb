Gemini: Here's a breakdown of why you might **have gotten a lower score** on the data after filling missing values and dropping rows:

**Information Loss:** When you fill missing values, even with a reasonable strategy like using the mean, you are essentially imputing data that isn't real. This can introduce noise or reduce the variability in the data, potentially making it harder for the model to learn meaningful patterns.

**Reduced Dataset Size:** By dropping rows with missing "Price" values, you reduced the size of your dataset. Machine learning models generally perform better with more data, as it allows them to learn more robust patterns. With less data, the model might not be able to generalize as well to unseen data.

**Nature of the Missing Data:** The missing data might not be missing randomly. There could be a pattern to why certain values are missing. If you simply fill or drop these values without understanding the underlying reason, you might be distorting the relationships within the data.

**Model Sensitivity:** The RandomForestRegressor model might be sensitive to the imputed values or the reduced dataset size. Different models might react differently to these data cleaning techniques.

## Extension: Feature Scaling

Copilot:

🔎 **Does AdaBoostRegressor need feature scaling (normalization/standardization)?**

*   AdaBoost is an ensemble method that builds a strong regressor from many weak learners (by default, decision trees).

*   Decision trees (CART) are the usual base learners for AdaBoost. Trees work by splitting on feature thresholds (e.g., x1 < 5.2).

*   Because splits are based on feature ordering, not on the absolute values, trees are invariant to monotonic transformations like scaling, standardization, or normalization.

👉 Therefore:

*   If you’re using trees as base estimators (the default), feature scaling is not necessary.

*   If you replace the base estimator with something sensitive to feature magnitudes (e.g., linear regression, SVM, neural nets), scaling becomes very important.

**Option 2: Fill missing values with Scikit-Learn**




```python
car_sales_missing = pd.read_csv("https://raw.githubusercontent.com/mrdbourke/zero-to-mastery-ml/refs/heads/master/data/car-sales-extended-missing-data.csv")
car_sales_missing.head()
```

```python
car_sales_missing.isna().sum()
```

```python
# Drop the rows with no labels
car_sales_missing.dropna(subset=["Price"], inplace = True)
car_sales_missing.isna().sum()
```

```python
# Split into x & y
X = car_sales_missing.drop("Price", axis = 1)
y = car_sales_missing["Price"]
```

```python
X.isna().sum()
```

```python
# Fill missing values with Scikil_learn
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

# Fill categrical values with 'missing' & numerical values with mean
cat_imputer = SimpleImputer(strategy='constant', fill_value='missing')
door_imputer = SimpleImputer(strategy='constant', fill_value=4)
num_imputer = SimpleImputer(strategy='mean')

# Define columns
cat_features = ["Make", "Colour"]
door_feature = ["Doors"]
num_features = ["Odometer (KM)"]

# Create an imputer (something that fills missing data)
imputer = ColumnTransformer(
    [
        ('cat_imputer', cat_imputer, cat_features),
        ('door_imputer', door_imputer, door_feature),
        ('num_imputer', num_imputer, num_features)
    ]
)

# Transform data
filled_X = imputer.fit_transform(X)
filled_X
```

```python
car_sales_filled = pd.DataFrame(filled_X,
                                columns=['Make', "Colour", 'Doors', 'Odometer (KM)'])
car_sales_filled.head()
```

```python
car_sales_filled.isna().sum()
```

```python
# Let's try and convert our data to numbers
# Turn the categories into numbers
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

categorical_features = ['Make', "Colour", "Doors"]
one_hot = OneHotEncoder()
transformer = ColumnTransformer([
    ('one_hot', one_hot, categorical_features)
], remainder='passthrough')

transformed_X = transformer.fit_transform(car_sales_filled)
transformed_X
```

```python
# Now we've got out data as numers and filled (no missing values)
# et's fit a model

np.random.seed(42)
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(transformed_X,
                                                    y,
                                                    test_size=0.2)
model = RandomForestRegressor()
model.fit(x_train, y_train)
model.score(x_test, y_test)
```
