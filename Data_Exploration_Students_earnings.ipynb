```python
import pandas as pd
```

```python
student_data = pd.read_csv('salaries_by_college_major.csv')
student_data.head()
```

```python
# How many rows are there?
student_data.shape
```

```python
# Data cleaning
student_data.isna().sum()
```

```python
clean_student_data = student_data.dropna()
clean_student_data.isna().sum()
```

```python
# Find College Major with Highest Starting Salaries
max_salary = clean_student_data['Starting Median Salary'].max()
id_max_salary = clean_student_data['Starting Median Salary'].idxmax()
job_max_salaray = clean_student_data["Undergraduate Major"][id_max_salary]
max_salary, job_max_salaray
```

```python
clean_student_data.loc[id_max_salary]
```

```python
# What college major has the highest mid-career salary?

id_max_mid = clean_student_data["Mid-Career Median Salary"].idxmax()
clean_student_data.loc[id_max_mid]
```

```python
# Which college major has the lowest starting salary and how much do graduates earn after university?

id_min_start = clean_student_data["Starting Median Salary"].idxmax()
mid_starting = clean_student_data.loc[id_min_start]
collage_major = mid_starting["Undergraduate Major"]
min_starting = mid_starting["Mid-Career Median Salary"]
collage_major, min_starting
```

```python
# Which college major has the lowest mid-career salary and how much can people expect to earn with this degree?

low_mid = clean_student_data["Mid-Career Median Salary"].idxmin()
mid_low = clean_student_data["Undergraduate Major"].loc[low_mid]
mid_low
```

```python
# Lowest Risk Majors
risk = clean_student_data["Mid-Career 90th Percentile Salary"] - clean_student_data["Mid-Career 10th Percentile Salary"]
# Using substract
risk_sub = clean_student_data["Mid-Career 90th Percentile Salary"].subtract(clean_student_data["Mid-Career 10th Percentile Salary"])
risk.head(), risk_sub.head()
```

```python
clean_student_data.insert(1, "Risk", risk)
clean_student_data.head()
```

```python
# Sorting by the Lowest Spread
sorted_risk = clean_student_data.sort_values("Risk")
sorted_risk[["Undergraduate Major", "Risk"]].head()
```

```python
# Sorting highest risk
desc_sorted_risk = clean_student_data.sort_values("Risk", ascending=False)
desc_sorted_risk[["Undergraduate Major", "Risk"]].head()
```

```python
# Highest potential
high_pot = clean_student_data.sort_values("Mid-Career 90th Percentile Salary", ascending=False)
high_pot[["Undergraduate Major", "Mid-Career 90th Percentile Salary"]].head()
```

```python
# Grouping and Pivoting Data
count_group = clean_student_data.groupby("Group").count()
count_group
```

```python
clean_student_data.head()
```

```python
clean_student_data.groupby("Group")[["Starting Median Salary", "Mid-Career Median Salary", "Mid-Career 10th Percentile Salary", "Mid-Career 90th Percentile Salary", "Risk"]].mean()
```
